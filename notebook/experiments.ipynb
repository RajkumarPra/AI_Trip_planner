```python
import os
import sys
from dotenv import load_dotenv

# 1. Add the parent directory to the system path to allow imports from your project
sys.path.append(os.path.abspath('..'))

# 2. Ensure environment variables are loaded (specifically OPENWEATHERMAP_API_KEY)
load_dotenv(dotenv_path='../.env') 

# 3. Import your tool class
from tools.weather_info_tool import WeatherInfoTool

# 4. Initialize the tool
weather_agent = WeatherInfoTool()
tools_list = weather_agent.weather_tool_list

# Langchain tool decorators create Tool objects. Let's find them:
current_weather_tool = next(t for t in tools_list if t.name == "get_current_weather")
forecast_weather_tool = next(t for t in tools_list if t.name == "get_weather_forecast")

city_to_test = "hyderabad"

print("🌤️ --- Testing Current Weather ---")
try:
    # We use .invoke() for Langchain tools
    result = current_weather_tool.invoke({"city": city_to_test})
    print(result)
except Exception as e:
    print(f"❌ Error fetching current weather: {e}")

print("\n📅 --- Testing Weather Forecast ---")
try:
    forecast_result = forecast_weather_tool.invoke({"city": city_to_test})
    print(forecast_result)
except Exception as e:
    print(f"❌ Error fetching forecast: {e}")
```
